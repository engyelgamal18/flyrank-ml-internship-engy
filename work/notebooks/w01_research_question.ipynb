## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 pages of these 16,262 pages (54.2%) are currently labelled as declining. There are also 6,561 pages that are both stale and still visible meaning they have not been updated for at least 180 days but still receive at least 500 impressions. These numbers suggest that there are many possible review candidates so ranking pages by priority could help a reviewer focus on the most useful ones first.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/engyelgamal18/flyrank-ml-internship-engy/main/data/raw/content_refresh_anonymized.csv"
df =pd.read_csv(url)

total_pages = len(df)

declining_pages = (df["trend_direction"]=="down").sum()
declining_pct = declining_pages / total_pages * 100

stale_visible = (
    (df["days_since_last_update"] >= 100) &
    (df["impressions_90d"] >= 500)
).sum()

print("Total pages:", total_pages)
print(f"Declining pages: {declining_pages} ({declining_pct:.1f}%)")
print("Stale + visible pages:", stale_visible)


```
